```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#  Data
df=sns.load_dataset('titanic')
```

```python
# Checking the data
df.head()
```

```python
# information of data
df.info()
```

# **Now Making The Plot**

```python
# scatter plot
plt.figure(figsize=(10,7))
sns.scatterplot(x='age',y='fare',data=df,hue='class',)# Hue is grouping the data based on class and showing different colors for each class in the scatter plot.
```

```python
# change the scatter piont color 
plt.figure(figsize=(10,7))
palette = {'male': 'red','female':'green'}
sns.scatterplot(x='age',y='fare',data=df,hue='sex',palette=palette)
# change the color of scatter point

#labele
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title("Scatter plot of age vs fare  with gender")
# saving the plot
plt.savefig('scatter_plot.png',dpi=320)
#saving svg file
plt.savefig('scatter_plot.svg',dpi=320)
```

**Style of plot**

```python
sns.set_theme(style='darkgrid')
figsize=(10,7)
plt.figure(figsize=figsize)
palette = {'male': 'red','female':'green'}
sns.scatterplot(x='age',y='fare',data=df,hue='sex',palette=palette)
plt.savefig('scatter_plot.svg',dpi=320)
```

# **Box Plot**


```python

sns.set_theme(style='darkgrid')
figsize=(10,7)
plt.figure(figsize=figsize)
palette = {'male': 'red','female':'green'}
sns.boxplot(x='sex',y='age',data=df,hue='sex',palette=palette)

plt.xlabel('Sex')
plt.ylabel('Age')
plt.title('Box plot of age by sex')
plt.savefig('box_plot.svg',dpi=320)
```

```python

sns.set_theme(style='darkgrid')
figsize=(10,7)
plt.figure(figsize=figsize)
palette = {'First': 'red','Second':'green','Third':'blue'}
sns.boxplot(x='sex',y='age',data=df,hue='class',palette=palette)

plt.xlabel('class')
plt.ylabel('Age')
plt.title('Box plot of age by sex')
plt.savefig('box_plot.svg',dpi=320)
```
