Stacked bar chart (Died + Survived)

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

# Stacked bar: pehle Died, phir Survived upar stack
plt.bar(result.index, result['Died'], label='Died')
plt.bar(result.index, result['Survived'], bottom=result['Died'], label='Survived')

plt.xlabel('Age Group')
plt.ylabel('Number of Passengers')
plt.title('Survival by Age Group (Titanic)')
plt.legend()
plt.grid(axis='y', alpha=0.3)
```

Side-by-side bars

```python
plt.figure(figsize=(10, 6))

width = 0.35
x = range(len(result))

plt.bar([i - width/2 for i in x], result['Died'], width, label='Died')
plt.bar([i + width/2 for i in x], result['Survived'], width, label='Survived')

plt.xticks(x, result.index)
plt.xlabel('Age Group')
plt.ylabel('Number of Passengers')
plt.title('Survival by Age Group (Titanic)')
plt.legend()
plt.grid(axis='y', alpha=0.3)

plt.show()
```
