Convert Whole Size Columns into bytes

```python
def convert_size(Size):
    if isinstance(Size,str):
        if 'k' in Size:
            return float(Size.replace('k',''))*1024
        elif 'M' in Size:
            return float(Size.replace("M",""))*1024 *1024
        elif 'Varies with device' in Size:
            return np.nan
    return Size
    
```

```python
df['Size']
```

```python
#let apply the function
df['Size'] = df['Size'].apply(convert_size)
```

# **Now apply Three opration in Installs Columns**
 1 ***Remove + Sing***

 2 ***Remove (,)***

```python
df['Installs']=df['Installs'].apply(lambda x:x.replace('+', '')if '+' in str(x) else x)
```

```python
df.head()
```

```python
# To remove the comma
df['Installs'] = df['Installs'].apply(lambda x: x.replace(',','')if ","in str(x) else x)
```

```python
df['Installs'].value_counts()
```

# **BINNING**

```python
# we cheak the maximum installs app
df['Installs'].max()
```

```python
bin =[-1,0,10,1000,10000,100000,1000000,10000000,1000000000,10000000000]
label = ['No','Very Low','Low','Moderate','More than Moderate','High','Very High','Very Very High','Top Notch']
df['install_catagory'] = pd.cut(df['Installs'],bins=bin,labels=label)
```

```python
#Cheak the value count of new columns
df['install_catagory'].value_counts()
```

```python
# convert Columns into integer 
df['Installs'] = df["Installs"].apply(lambda x: int(x))
```

```python
#make figure size
plt.figure(figsize=(16,6))

#Plot the null values by their percentage in each columns
missing_percentage = df.isnull().sum()/len(df)*100

missing_percentage.plot(kind='bar')

#add the labels
plt.xlabel('columns')
plt.ylabel('Percentage')
plt.title('Percentage of Missing Values in each columns')

plt.ylim(0, 16)
```

```python
# those value which missing percentage in less then 1 percent
missing_percentage[missing_percentage < 1].plot(kind='bar')
```
